## 模型调用（Model Invocation）

本教程演示如何使用 LangChain 调用大语言模型（LLM），包括：
1. 直接调用模型生成文本
2. 使用 ChatModel 进行对话
3. 流式输出（Streaming）
4. 使用 Prompt Template

### 1. 安装和导入依赖

```python
# 安装必要的包
# !pip install langchain langchain-openai langchain-anthropic langchain-community

import os
from langchain_openai import ChatOpenAI, OpenAI
from langchain_core.prompts import PromptTemplate
import dotenv

dotenv.load_dotenv()
# 设置 API Key（如果不在环境变量中）
# os.environ["OPENAI_API_KEY"] = "your-api-key-here"
# os.environ["ANTHROPIC_API_KEY"] = "your-api-key-here"
```

### 2. 直接调用 LLM（Completion Model）

使用 `OpenAI` 类调用传统的补全模型：

```python
# 创建 LLM 实例
llm = ChatOpenAI(
    model=os.getenv("MODEL_NAME"),
    openai_api_key=os.getenv("OPENAI_API_KEY"),
    openai_api_base=os.getenv("OPENAI_BASE_URL"),
)

# 直接调用，传入字符串
try:
    response = llm.invoke(input="什么是机器学习？")
except ValueError as e:
    print(f"err:{e}")
finally:
    print("end")
    
print(response)
```

### 3. 使用 ChatModel（对话模型）

ChatModel 更适合对话场景，支持系统消息、用户消息等：

```python
from langchain.schema import HumanMessage, SystemMessage, AIMessage

# 使用相同的配置
chat = ChatOpenAI(
    model=os.getenv("MODEL_NAME", "gpt-3.5-turbo"),
    openai_api_key=os.getenv("OPENAI_API_KEY"),
    openai_api_base=os.getenv("OPENAI_BASE_URL"),
    temperature=0.7,
)

# 方式1：简单调用
response = chat.invoke("你好，请介绍一下自己")
print(response.content)

# 方式2：使用消息列表（推荐）
messages = [
    SystemMessage(content="你是一个专业的Python编程助手"),
    HumanMessage(content="解释一下什么是装饰器"),
]
response = chat.invoke(messages)
print(response.content)
```

### 4. 流式输出（Streaming）

逐token输出，适合实时展示长文本生成：

```python
# 流式输出
chat = ChatOpenAI(model="gpt-3.5-turbo", streaming=True)

# 使用 stream 方法
for chunk in chat.stream("用Python写一个快速排序算法"):
    print(chunk.content, end="", flush=True)
```

### 5. 使用 PromptTemplate

模板化提示词，方便复用和管理：

```python
from langchain.prompts import ChatPromptTemplate

# 创建提示模板
prompt = ChatPromptTemplate.from_messages([
    ("system", "你是一位{field}领域的专家"),
    ("human", "请解释{concept}是什么？"),
])

# 使用模板
chain = prompt | chat

# 调用
response = chain.invoke({
    "field": "人工智能",
    "concept": "神经网络"
})
print(response.content)
```

### 6. 使用 Anthropic Claude 模型

切换不同的模型提供商非常简单：

```python
# 使用 Claude 模型
claude = ChatAnthropic(
    model="claude-3-5-sonnet-20241022",
    temperature=0.7,
    max_tokens=1024,
)

response = claude.invoke("用Python写一个读取CSV文件的函数")
print(response.content)
```
