## 使用init_chat_model初始化大模型实例

### stream 流式输出（阻塞式）

```python
from langchain.chat_models import init_chat_model
from langchain_core.messages import SystemMessage, HumanMessage

llm = init_chat_model(model="qwen2.5:0.5b",model_provider="ollama")

messages = [
    SystemMessage(content="你是一个Python编程助手，回答要简洁。"),
    HumanMessage(content="Python中如何读取JSON文件？"),
]

for chunk in llm.stream(messages):
    print(chunk.content,end="",flush=True)

print("\nend")
```

### astream (非阻塞式输出)

```python
from langchain.chat_models import init_chat_model
from langchain_core.messages import SystemMessage, HumanMessage

llm = init_chat_model(model="qwen2.5:0.5b",model_provider="ollama")

messages = [
    SystemMessage(content="你是一个Python编程助手，回答要简洁。"),
    HumanMessage(content="Python中如何读取JSON文件？"),
]

async for chunk in llm.astream(messages):
    print(chunk.content,end="",flush=True)

print("\nend")
```

### invoke 阻塞模式调用

```python
from langchain_ollama import ChatOllama
from langchain_core.messages import SystemMessage, HumanMessage

llm = ChatOllama(model="qwen2.5:0.5b")

messages = [
    SystemMessage(content="你是一个Python编程助手，回答要简洁。"),
    HumanMessage(content="Python中如何读取JSON文件？"),
]

response = llm.invoke(messages)
print(response.content)
```

### ainvoke 非阻塞调用

```python
from langchain_ollama import ChatOllama
from langchain_core.messages import SystemMessage, HumanMessage

llm = ChatOllama(model="qwen2.5:0.5b")

messages = [
    SystemMessage(content="你是一个Python编程助手，回答要简洁。"),
    HumanMessage(content="Python中如何读取JSON文件？"),
]

# ainvoke 是 invoke 的异步版本，需要用 await 调用
response = await llm.ainvoke(messages)
print(response.content)
```

### batch 批量调用（同步）

```python
from langchain_ollama import ChatOllama
from langchain_core.messages import SystemMessage, HumanMessage

llm = ChatOllama(model="qwen2.5:0.5b")

messages1 = [
    SystemMessage(content="你是一个Python编程助手，回答要简洁。"),
    HumanMessage(content="Python中如何读取JSON文件？"),
]

messages2 = [
    SystemMessage(content="你是一个Python编程助手，回答要简洁。"),
    HumanMessage(content="什么是列表推导式？"),
]

# batch 批量调用，同时处理多个请求
responses = llm.batch([messages1, messages2])

for i, response in enumerate(responses):
    print(f"\n--- 回复 {i+1} ---")
    print(response.content)
    print()
```

### abatch 批量调用（异步）

```python
from langchain_ollama import ChatOllama
from langchain_core.messages import SystemMessage, HumanMessage

llm = ChatOllama(model="qwen2.5:0.5b")

messages1 = [
    SystemMessage(content="你是一个Python编程助手，回答要简洁。"),
    HumanMessage(content="Python中如何读取JSON文件？"),
]

messages2 = [
    SystemMessage(content="你是一个Python编程助手，回答要简洁。"),
    HumanMessage(content="什么是列表推导式？"),
]

# abatch 是 batch 的异步版本，需要用 await 调用
responses = await llm.abatch([messages1, messages2])

for i, response in enumerate(responses):
    print(f"--- 回复 {i+1} ---")
    print(response.content)
    print()
```
