# 提示词模板 prompt template

## 1. PromptTemplate 的使用

### 使用 PromptTemplate构造方法
1. 使用format方法

```python
from langchain_core.prompts import PromptTemplate

# 1. 创建实例
prompt_template = PromptTemplate(
    template="请用简单的语言解释{topic}。",
    input_variables=["topic"],
)

# 2. 填充变量
prompt = prompt_template.format(topic="机器学习")

print(prompt)
```

2. 使用form_template方法（推荐）

```python
from langchain_core.prompts import PromptTemplate

# 1. 创建实例
prompt_template = PromptTemplate.from_template(
    template="请用简单的语言解释{topic}。",
    # input_variables=["topic"],
)

# 2. 填充变量
prompt = prompt_template.format(topic="机器学习")

print(prompt)
```

### 部分变量（默认值）


```python
from langchain_core.prompts import PromptTemplate

# 1. 创建实例
prompt_template = PromptTemplate.from_template(
    template="请用简单的语言解释{topic}。并且{extra_info}",
    # input_variables=["topic"],
    partial_variables={"topic": "机器学习"}, # 部分变量，相当于默认值
)

# 2. 填充变量
prompt = prompt_template.format(extra_info="请用中文回答",topic="深度学习")

print(prompt)
```

还可以使用partial函数

```python
from langchain_core.prompts import PromptTemplate

# 1. 创建实例
prompt_template = PromptTemplate(
    template="请用简单的语言解释{topic}。并且用{language}回答。",
    input_variables=["topic","language"],
)

# prompt_template.partial 返回新的一个实例
prompt_template_new = prompt_template.partial(topic="机器学习")

# 2. 填充变量
prompt = prompt_template_new.format(language="中文")

print(prompt)
```

2. 使用format方法

```python
from langchain_core.prompts import PromptTemplate

# 1. 创建实例
prompt_template = PromptTemplate.from_template(
    template="请用简单的语言解释{topic}。并且{extra_info}",
    # input_variables=["topic"],
    partial_variables={"topic": "机器学习"}, # 部分变量，相当于默认值
)

# 2. 填充变量
prompt = prompt_template.invoke({"extra_info":"请用中文回答","topic":"深度学习"})

print(prompt)
print(type(prompt))
```

总结：invoke使用的是字典,返回值为StringPromptValue；而format函数使用字符串，返回值为str
