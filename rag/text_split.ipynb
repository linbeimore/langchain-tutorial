### 2.1 CharacterTextSplitter
> 字符分割器，按字符长度进行分割

```python
from langchain_text_splitters import CharacterTextSplitter
from rich import print as rprint

text = """
2026年国际足联世界杯（英语：2026 FIFA World Cup Canada/Mexico/United States，西班牙语：Copa mundial de la FIFA Canadá/México/Estados Unidos 2026，法语：Coupe du monde de la FIFA de 2026 Canada/Mexique/États-Unis）为第23届国际足联世界杯赛事，定于2026年6月11日至7月19日举行，赛事在美国、墨西哥、加拿大共16座城市展开，为世界杯首次由3国联合举办。

本届世界杯是继2002年赛事后再次由多国联合主办，第4次由中北美洲及加勒比海地区主办。墨西哥此前曾举办1970年和1986年世界杯，本届赛事后成为首个3办世界杯之国；美国曾于1994年举办赛事；加拿大则是首次举办男子世界杯，恰巧的是三者均在本届世界杯淘汰赛16强出局。此外，本届赛事将回归传统的北半球夏季档期。

本届赛事首次由过往的32支参赛球队扩大至48支。作为主办国，三个东道主自动获得决赛周参赛资格。在入围名单中，佛得角、库拉索、约旦和乌兹别克斯坦均为首次晋身世界杯决赛周。本届赛事的卫冕冠军为阿根廷。
"""

splitter = CharacterTextSplitter(chunk_size=200, chunk_overlap=50,separator="。")

texts = splitter.split_text(text)

rprint(f"[bold]共 {len(texts)} 个片段[/bold]")
rprint(texts)

for text in texts:
    rprint("-" * 20)
    rprint(f"[bold]片段长度：{len(text)}[/bold]")
    rprint(text)
```

### 2.2 RecursiveCharacterTextSplitter
> 递归字符串切分

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
from rich import print as rprint

text = """2026年国际足联世界杯（英语：2026 FIFA World Cup Canada/Mexico/United States，西班牙语：Copa mundial de la FIFA Canadá/México/Estados Unidos 2026，法语：Coupe du monde de la FIFA de 2026 Canada/Mexique/États-Unis）为第23届国际足联世界杯赛事，定于2026年6月11日至7月19日举行，赛事在美国、墨西哥、加拿大共16座城市展开，为世界杯首次由3国联合举办。

本届世界杯是继2002年赛事后再次由多国联合主办，第4次由中北美洲及加勒比海地区主办。墨西哥此前曾举办1970年和1986年世界杯，本届赛事后成为首个3办世界杯之国；美国曾于1994年举办赛事；加拿大则是首次举办男子世界杯，恰巧的是三者均在本届世界杯淘汰赛16强出局。此外，本届赛事将回归传统的北半球夏季档期。

本届赛事首次由过往的32支参赛球队扩大至48支。作为主办国，三个东道主自动获得决赛周参赛资格。在入围名单中，佛得角、库拉索、约旦和乌兹别克斯坦均为首次晋身世界杯决赛周。本届赛事的卫冕冠军为阿根廷。
"""

list = [text,text]

splitter = RecursiveCharacterTextSplitter(chunk_size=200, chunk_overlap=50,add_start_index=True, separators=["\n\n","。"])

doc_list = splitter.create_documents(list)


for doc in doc_list:
    rprint("-" * 20)
    rprint(doc)
```

```python
from langchain_text_splitters import CharacterTextSplitter
from rich import print as rprint

text = """2026年国际足联世界杯（英语：2026 FIFA World Cup Canada/Mexico/United States，西班牙语：Copa mundial de la FIFA Canadá/México/Estados Unidos 2026，法语：Coupe du monde de la FIFA de 2026 Canada/Mexique/États-Unis）为第23届国际足联世界杯赛事，定于2026年6月11日至7月19日举行，赛事在美国、墨西哥、加拿大共16座城市展开，为世界杯首次由3国联合举办。

本届世界杯是继2002年赛事后再次由多国联合主办，第4次由中北美洲及加勒比海地区主办。墨西哥此前曾举办1970年和1986年世界杯，本届赛事后成为首个3办世界杯之国；美国曾于1994年举办赛事；加拿大则是首次举办男子世界杯，恰巧的是三者均在本届世界杯淘汰赛16强出局。此外，本届赛事将回归传统的北半球夏季档期。

本届赛事首次由过往的32支参赛球队扩大至48支。作为主办国，三个东道主自动获得决赛周参赛资格。在入围名单中，佛得角、库拉索、约旦和乌兹别克斯坦均为首次晋身世界杯决赛周。本届赛事的卫冕冠军为阿根廷。
"""

list = [text]

splitter = CharacterTextSplitter.from_tiktoken_encoder(chunk_size=200, chunk_overlap=0,encoding_name="cl100k_base")

doc_list = splitter.create_documents(list)


for doc in doc_list:
    rprint("-" * 20)
    rprint(doc)
```
