## textloader

> Document Loader 用于将不同格式的文件加载为 LangChain 的 `Document` 对象。
> 每个 Document 包含 `page_content`（文本内容）和 `metadata`（元数据：文件名、来源等）。

### mineru
> 相比 PyPDFLoader，MinerU 支持公式识别、表格识别、OCR，对复杂 PDF 的解析效果更好。

```python
import os
import time
import requests
from rich import print as rprint
from dotenv import load_dotenv

load_dotenv(override=True)

# ══════════════════════════════════════════════════════════════
# MinerU API：将 PDF/图片/Word 等文档解析为 Markdown
# 文档：https://mineru.net/apiManage/docs
# 流程：获取上传链接 → 上传文件 → 轮询状态 → 下载结果
# ══════════════════════════════════════════════════════════════

MINERU_TOKEN = os.getenv("MINERU_TOKEN")
BASE_URL = "https://mineru.net"


def get_upload_urls(file_paths: list[str]) -> dict:
    """
    步骤1：批量获取签名上传链接
    POST /api/v4/file-urls/batch
    返回 batch_id 和每个文件的上传 URL
    """
    url = f"{BASE_URL}/api/v4/file-urls/batch"
    headers = {
        "Authorization": f"Bearer {MINERU_TOKEN}",
        "Content-Type": "application/json",
    }

    files_info = [
        {
            "name": os.path.basename(fp),
            "is_ocr": True,
            "data_id": f"file_{i}",
        }
        for i, fp in enumerate(file_paths)
    ]

    data = {
        "files": files_info,
        "model_version": "vlm",       # vlm 精度更高
        "enable_formula": True,        # 启用公式识别
        "enable_table": True,          # 启用表格识别
        "language": "ch",              # 中文 OCR
    }

    resp = requests.post(url, headers=headers, json=data)
    if resp.status_code != 200:
        raise Exception(f"获取上传链接失败：{resp.status_code} {resp.text}")

    result = resp.json()
    if result.get("code") != 0:
        raise Exception(f"API 错误：{result.get('msg')}")

    return result["data"]


def upload_to_oss(file_path: str, upload_url: str):
    """
    步骤2：将本地文件 PUT 到签名 URL
    上传完成后，MinerU 自动开始解析
    """
    with open(file_path, "rb") as f:
        resp = requests.put(upload_url, data=f)
    if resp.status_code not in (200, 201):
        raise Exception(f"上传文件失败：{resp.status_code}")
    rprint(f"  ✅ 上传成功：{os.path.basename(file_path)}")


def poll_batch_status(batch_id: str, max_wait: int = 300, interval: int = 5) -> list[dict]:
    """
    步骤3：轮询批量任务状态
    GET /api/v4/extract-results/batch/{batch_id}
    """
    url = f"{BASE_URL}/api/v4/extract-results/batch/{batch_id}"
    headers = {"Authorization": f"Bearer {MINERU_TOKEN}"}

    elapsed = 0
    while elapsed < max_wait:
        resp = requests.get(url, headers=headers)
        result = resp.json()

        if result.get("code") != 0:
            raise Exception(f"查询失败：{result.get('msg')}")

        extract_results = result["data"]["extract_result"]
        states = [r["state"] for r in extract_results]

        rprint(f"  ⏳ [{elapsed}s] 状态：{states}")

        if all(s == "done" for s in states):
            rprint("  ✅ 全部完成！")
            return extract_results
        if any(s == "failed" for s in states):
            failed = [r for r in extract_results if r["state"] == "failed"]
            raise Exception(f"部分任务失败：{failed}")

        time.sleep(interval)
        elapsed += interval

    raise Exception(f"超时（{max_wait}s），任务未完成")


def download_results(results: list[dict], output_dir: str = "./mineru_output"):
    """
    步骤4：下载解析结果（ZIP 包含 Markdown + JSON）
    """
    os.makedirs(output_dir, exist_ok=True)

    for i, r in enumerate(results):
        zip_url = r.get("full_zip_url")
        file_name = r.get("file_name", f"result_{i}.zip")
        if not zip_url:
            continue

        rprint(f"  📥 下载：{file_name}")
        resp = requests.get(zip_url)
        zip_path = os.path.join(output_dir, file_name.replace(".pdf", ".zip"))
        with open(zip_path, "wb") as f:
            f.write(resp.content)
        rprint(f"  💾 保存：{zip_path}")


# ══════════════════════════════════════════════════════════════
# 完整示例：解析一个本地 PDF
# ══════════════════════════════════════════════════════════════

file_paths = ["../asset/load/read.pdf"]

rprint("[bold]🚀 步骤1：获取上传链接[/bold]")
upload_data = get_upload_urls(file_paths)
batch_id = upload_data["batch_id"]
upload_urls = upload_data["file_urls"]
rprint(f"  batch_id: {batch_id}")

rprint("\n[bold]📤 步骤2：上传文件到 OSS[/bold]")
for fp, url in zip(file_paths, upload_urls):
    upload_to_oss(fp, url)

rprint("\n[bold]⏳ 步骤3：轮询解析状态[/bold]")
results = poll_batch_status(batch_id)

rprint("\n[bold]📥 步骤4：下载结果[/bold]")
download_results(results)

rprint("\n[bold green]🎉 MinerU 解析完成！[/bold green]")
```

## 9. DirectoryLoader
> 批量加载一个文件夹内所有文件

```python
import os
from langchain_community.document_loaders import (
    DirectoryLoader,
    TextLoader,
    CSVLoader,
    UnstructuredHTMLLoader,
    UnstructuredMarkdownLoader,
)
from rich import print as rprint

load_dir = "../asset/load"

# ══════════════════════════════════════════════════════════════
# 方式一：加载单一类型文件（指定 loader_cls）
# ══════════════════════════════════════════════════════════════
rprint("[bold]方式一：加载所有 .txt 文件[/bold]")
txt_loader = DirectoryLoader(
    load_dir,
    glob="**/*.txt",
    loader_cls=TextLoader,
    loader_kwargs={"encoding": "utf-8"},
    show_progress=True,
)
txt_docs = txt_loader.load()
rprint(f"  找到 {len(txt_docs)} 个 txt 文件\n")

# ══════════════════════════════════════════════════════════════
# 方式二：加载混合类型文件（指定 loader_cls 为对应类型）
# ══════════════════════════════════════════════════════════════
rprint("[bold]方式二：加载所有 .html 文件[/bold]")
html_loader = DirectoryLoader(
    load_dir,
    glob="**/*.html",
    loader_cls=UnstructuredHTMLLoader,
    loader_kwargs={"mode": "elements", "strategy": "hi_res"},
    show_progress=True,
)
html_docs = html_loader.load()
rprint(f"  找到 {len(html_docs)} 个 html 文件\n")

# ══════════════════════════════════════════════════════════════
# 方式三：使用 silent_errors 忽略加载失败的文件
# ══════════════════════════════════════════════════════════════
rprint("[bold]方式三：加载所有文件，忽略错误[/bold]")
all_loader = DirectoryLoader(
    load_dir,
    glob="**/*.*",
    loader_cls=TextLoader,
    loader_kwargs={"encoding": "utf-8"},
    silent_errors=True,      # 加载失败的文件不中断，跳过并记录
    show_progress=True,
)
all_docs = all_loader.load()
rprint(f"  共加载 {len(all_docs)} 个文档\n")

# 展示结果
rprint("[bold]加载结果汇总：[/bold]")
for doc in all_docs:
    source = doc.metadata.get("source", "unknown")
    filename = os.path.basename(source)
    preview = doc.page_content[:60].replace("\n", " ")
    rprint(f"  📄 {filename}: {preview}...")
```
